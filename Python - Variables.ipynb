### ცვლადები და მონაცემთა ტიპები

ცვლადი ინახავს გარკვეულ ინფორმაციას. პითონში ცვლადის დასახელება უნდა იწყებოდეს ანბანის ასოთი ან ქვედა ტირით („_“) და შეიძლება შეიცავდეს ანბანის ასოებს, ციფრებს და ქვედა ტირეს. ცვლადში შენახული მონაცებემი განსხვავდება ტიპის მიხედვით.

მაგალითად, შევქმნათ ცვლადი, რომელიც შეიცავს ტექსტს:

```python
a = "Hello World"
```

```python
print(a)
```

პითონს ახასიათებს დინამიური ტიპიზაცია - ცვლადის ტიპი განისაზღვრება მისთვის მინიჭებული მნიშვნელობით. მაგალითად, თუ ცვლადს მივანიჭებთ ერთმაგ ან ორმაგ ბრჭყალებში ჩასმულ სიმბოლოების თანმიმდევრობას, როგორც ზემოთაა მოცემული, მისი ტიპი გახდება string ანუ str. მთელი რიცხვითი მნიშვნელობის მინიჭებისას, პითონი ცვლადს ავტომატურად ანიჭებს integer ანუ int ტიპს. 

```python
b = 1
```

```python
print(b)
```

როგორც ვნახეთ, ცვლადებს მნიშვნელობას ვანიჭებთ = სიმბოლოს მეშვეობით. რა მოხდება, თუ ცვლადს თავიდან მივანიჭებთ მნიშვნელობას?

```python
a = b
```

```python
print(a)
```

როგორც ხედავთ, a ცვლადის მნიშვნელობა შეიცვალა. ეს მოხდა იმიტომ, რომ a ცვლადს მივანიჭეთ b-ში შენახული მნიშვნელობა, რომელიც მინიჭების მომენტში იყო 1.

```python
print(b)
```

```python
a = "Hello World"
b = 1
```

```python
a = b
b = a
```

```python
print(a)
print(b)
```

b = a ოპერაციით ჩვენ b-ს a ცვლადში შენახული მნიშვნელობა მივანიჭეთ, რომელიც უკვე 1 იყო, რადგან იქამდე შესრულდა a = b.

```python
a = "Hello World"
b = 1
```

იმისათვის რომ a-ს მნიშვნელობა შევცვალოთ ისე რომ არ დავკგარგოთ მისი ამჟამინდელი მნიშვნელობა, შეგვიძლია დავიხმაროთ დროებითი ცვლადები:

```python
temp1 = a
temp2 = b
```

```python
print(temp1)
print(temp2)
print(a)
print(b)
```

```python
a = temp2
b = temp1
```

```python
print(a)
print(b)
```
